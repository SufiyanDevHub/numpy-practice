```python
#Aggregation operation in numpy
```

```python
arr = np.array([10,20,30,40])
print(arr.sum())# return the total sum of an array
print(arr.mean())# return the  mean(average) of array elements
print(arr.max()) # return the largest element of an array
print(arr.min()) # return the smallest value of an array
print(arr.std())
print(arr.var())
```

```python
print((10+20+30+40)/4)
```

```python
arr_st = np.array([22,24,26,28])
print(arr_st.mean())
print(arr_st.std())
```

```python
arr_n = np.array([
    [1,2,3],
    [4,5,6]
])
print(arr_n.mean())
```

```python
print((1+2+3+4+5+6)/6)
```

```python
# When we want the fixed colum or row operation, we use axis
```

```python
print(arr_n.sum(axis=0))
print(arr_n.sum(axis=1))
print(arr_n.mean(axis=1))
print(arr_n.min(axis=1))
print(arr_n.max(axis=1))
print(arr_n.std(axis=1))
print(arr_n.var(axis=1))
```

```python
#logical operator
```

```python
arr = np.array([10,20,30,40,50])
print(arr>25) # for check condintion and return true or false based on condition
print(arr[arr>25])
print(arr[arr<35])
print(arr[arr%2 == 2])
```

```python
print(arr[(arr>20)&(arr<50)])
print(arr[(arr>15)|(arr<45)])
```
